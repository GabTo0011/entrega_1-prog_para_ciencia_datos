```python
# Columnas de sexo
if {'sexopaciente', 'numtotal'}.issubset(df.columns):
    sexo_totales = df.groupby('sexopaciente')['numtotal'].sum().sort_values(ascending=False)
    
    plt.figure(figsize=(8, 5))
    sns.barplot(x=sexo_totales.index, y=sexo_totales.values, palette='plasma')
    plt.title('Total de Atenciones por Sexo', fontsize=16)
    plt.ylabel('Número de Atenciones', fontsize=12)
    plt.xlabel('Sexo del Paciente', fontsize=12)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

```python
if {'prioridadtriage', 'numtotal'}.issubset(df.columns):
    triage_totales = df.groupby('prioridadtriage')['numtotal'].sum().sort_index(ascending=True)
    
    plt.figure(figsize=(10, 6))
    sns.barplot(x=triage_totales.index, y=triage_totales.values, palette='cividis')
    plt.title('Distribución de Atenciones por Nivel de Triage', fontsize=16)
    plt.ylabel('Número de Atenciones', fontsize=12)
    plt.xlabel('Nivel de Triage (C1=Más grave, C5=Menos grave)', fontsize=12)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Gráfico por Región
if {'regionglosa', 'numtotal'}.issubset(df.columns):
    region_totales = df.groupby('regionglosa')['numtotal'].sum().sort_values(ascending=False)
    sns.barplot(ax=axes[0], x=region_totales.index, y=region_totales.values, palette='crest')
    axes[0].set_title('Atenciones por Región', fontsize=16)
    axes[0].set_ylabel('Número de Atenciones', fontsize=12)
    axes[0].set_xlabel('Región', fontsize=12)
    axes[0].tick_params(axis='x', rotation=45)

# Gráfico por Tipo de Establecimiento
if 'tipoestablecimiento' in df.columns:
    tipo_establecimiento_totales = df['tipoestablecimiento'].value_counts()
    axes[1].pie(tipo_establecimiento_totales, labels=tipo_establecimiento_totales.index, autopct='%1.1f%%',
              startangle=90, colors=sns.color_palette('pastel'))
    axes[1].set_title('Proporción de Atenciones por Tipo de Establecimiento', fontsize=16)
    axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
```
